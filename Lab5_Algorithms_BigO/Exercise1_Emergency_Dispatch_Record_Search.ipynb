```python
# EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH

# 1. Student-Specific Inputs
LAST_NAME = input("Enter your surname: ").strip()
STUDENT_ID = input("Enter your student ID number: ").strip()
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip().upper()

# 2. Generate Student-Specific Values
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))
ARTIST_SUM = sum(
    ord(letter) - ord("A") + 1
    for letter in FAVORITE_ARTIST
    if "A" <= letter <= "Z"
)

print(f"\n--- Student-Specific Values ---")
print(f"SEED_DIGIT: {SEED_DIGIT}")
print(f"ID_SUM: {ID_SUM}")
print(f"NAME_LENGTH: {NAME_LENGTH}")
print(f"ARTIST_LENGTH: {ARTIST_LENGTH}")
print(f"ARTIST_SUM: {ARTIST_SUM}")

# 3. Generate Original Incident Records
SIZE = 20
base = 1000 + ID_SUM * 100 + ARTIST_SUM
original = list(range(base, base + SIZE))

# 4. Rotate the Dataset Using SEED_DIGIT
rotation = SEED_DIGIT % SIZE
rotated = original[rotation:] + original[:rotation]

# 5. Generate Target Incident
target_index = (ID_SUM + ARTIST_SUM) % SIZE
target = original[target_index]


# 6. Rotated Binary Search
def rotated_binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons

        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1

    return -1, comparisons


# 7. Linear Search
def linear_search(data, target):
    comparisons = 0

    for index, value in enumerate(data):
        comparisons += 1

        if value == target:
            return index, comparisons

    return -1, comparisons


# 8. Search for the Target
binary_position, binary_comparisons = rotated_binary_search(
    rotated, target
)
linear_position, linear_comparisons = linear_search(
    rotated, target
)

# 9. Display Assessment Data
print(f"\n--- Assessment Data ---")
print(f"Original Dataset: {original}")
print(f"Rotated Dataset: {rotated}")
print(f"Target Incident: {target}")

if binary_position != -1:
    print(f"Binary Search Position: {binary_position}")
else:
    print("Binary Search: Target not found")

print(f"Binary Search Comparisons: {binary_comparisons}")

if linear_position != -1:
    print(f"Linear Search Position: {linear_position}")
else:
    print("Linear Search: Target not found")

print(f"Linear Search Comparisons: {linear_comparisons}")

print("\n--- Complexity ---")
print("Rotated Binary Search: O(log n) time, O(1) auxiliary space")
print("Linear Search: O(n) time, O(1) auxiliary space")

print("\n--- Final Output ---")
if binary_position == linear_position and binary_position != -1:
    print("Both methods found the target at the same position.")
else:
    print("Check the search results.")

print(f"Binary Search comparisons: {binary_comparisons}")
print(f"Linear Search comparisons: {linear_comparisons}")
```
