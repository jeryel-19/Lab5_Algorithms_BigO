```python
# EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS

# Student-specific values from Exercise 1
SEED_DIGIT = 6
ID_SUM = 16
ARTIST_LENGTH = 6
NAME_LENGTH = 6

# 1. Generate two sorted datasets
SIZE_A = 10
SIZE_B = 10

dataset_a = [
    SEED_DIGIT * 100 + ID_SUM + i * 2
    for i in range(SIZE_A)
]

dataset_b = [
    ARTIST_LENGTH * 100 + NAME_LENGTH + i * 3
    for i in range(SIZE_B)
]

# 2. Merge the sorted datasets and count comparisons
def merge_sorted(a, b):
    i = 0
    j = 0
    merged = []
    comparisons = 0

    while i < len(a) and j < len(b):
        comparisons += 1

        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1

    merged.extend(a[i:])
    merged.extend(b[j:])

    return merged, comparisons

consolidated, comparison_count = merge_sorted(
    dataset_a, dataset_b
)

# 3. Verify ordering and record counts
is_sorted = all(
    consolidated[i] <= consolidated[i + 1]
    for i in range(len(consolidated) - 1)
)

print(f"--- Assessment Data ---")
print(f"Department A dataset: {dataset_a}")
print(f"Department B dataset: {dataset_b}")
print(f"Consolidated dataset: {consolidated}")

print(f"Records before consolidation: {len(dataset_a)} + {len(dataset_b)} = {len(dataset_a) + len(dataset_b)}")
print(f"Records after consolidation: {len(consolidated)}")
print(f"Comparison count: {comparison_count}")
print(f"Ordering verification: {'Passed' if is_sorted else 'Failed'}")

print(f"Time Complexity: O(n + m)")
print(f"Auxiliary Space Complexity: O(n + m)")

print(f"\n--- Final Output ---")
if is_sorted and len(consolidated) == len(dataset_a) + len(dataset_b):
    print(f"Consolidation successful. All records are in order.")
else:
    print("Consolidation verification failed.")
```
