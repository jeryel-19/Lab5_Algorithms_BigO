```python
# EXERCISE 3: E-COMMERCE ORDER PROCESSING

import random
import time

# 1. Student-specific values
SEED_DIGIT = 6
ID_SUM = 16
NAME_LENGTH = 6
ARTIST_LENGTH = 6
ARTIST_SUM = 57

# 2. Generate the original order dataset
SIZE = 20
base = 1000 + ID_SUM * 10 + SEED_DIGIT

orders = [base + i * (NAME_LENGTH + 1) for i in range(SIZE)]

# Generate the three input conditions from the same orders
sorted_orders = orders.copy()
reverse_orders = orders[::-1]

random_orders = orders.copy()
random.Random(ID_SUM + ARTIST_SUM).shuffle(random_orders)

datasets = {
    "Sorted": sorted_orders,
    "Reverse-ordered": reverse_orders,
    "Random": random_orders
}

# 3. Bubble Sort
def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0
    n = len(arr)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 3
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements

# 4. Selection Sort
def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0
    n = len(arr)

    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 3

    return arr, comparisons, movements

# 5. Insertion Sort
def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements

# 6. Apply all algorithms and record results
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

print(f"--- Generated Order Dataset ---")
print(f"Original orders: {orders}")
print(f"Number of orders: {len(orders)}")

print(f"\n--- Input Conditions ---")
for condition, data in datasets.items():
    print(f"{condition}: {data}")

print(f"\n--- Performance Results ---")

for condition, data in datasets.items():
    print(f"\nInput Condition: {condition}")

    for name, sort_function in algorithms.items():
        start = time.perf_counter()
        result, comparisons, movements = sort_function(data)
        elapsed = time.perf_counter() - start

        verified = result == sorted(orders)

        print(f"\n{name}")
        print(f"Sorted result: {result}")
        print(f"Comparisons: {comparisons}")
        print(f"Movements: {movements}")
        print(f"Execution time: {elapsed:.8f} seconds")
        print(f"Sorting verification: {'Passed' if verified else 'Failed'}  ")

# 7. Big O analysis
print(f"\n--- Big O Analysis ---")
print(f"Bubble Sort: Best O(n), Average O(n^2), Worst O(n^2)")
print(f"Selection Sort: Best O(n^2), Average O(n^2), Worst O(n^2)")
print(f"Insertion Sort: Best O(n), Average O(n^2), Worst O(n^2)")
print(f"Auxiliary space: O(1) for all three algorithms")

print(f"\n--- Final Output ---")
print(f"All three sorting algorithms were tested under all three input conditions.")
print(f"Check the verification results to confirm that every output is sorted.")
```
